## 1. ¿Por qué descubrir un fármaco tarda 12 años?

Antes de escribir una sola línea de código, hagámonos la pregunta más importante del curso:

> **¿Cuánto tiempo y dinero cuesta llevar una molécula desde el laboratorio hasta la farmacia?**

La respuesta es frustrante: en promedio **10–15 años** y entre **1.000 y 2.600 millones de dólares** por fármaco aprobado.  
Y de cada 10.000 moléculas que entran al proceso, **solo 1** llega al mercado.

¿Por qué? Porque el espacio de moléculas posibles es astronómicamente grande.


````python
# ── Calculemos juntos la magnitud del problema ──────────────────────────────
# No se preocupen por la sintaxis — la entenderán al final de la semana.
# Por ahora, lean los comentarios y observen los resultados.

# Número estimado de moléculas drug-like posibles (con Mw < 500 Da)
espacio_quimico = 10**60

# Moléculas que tiene ChEMBL (la base de datos que usaremos en este curso)
moleculas_en_chembl = 2_400_000

# ¿Qué fracción del espacio químico cubre ChEMBL?
fraccion = moleculas_en_chembl / espacio_quimico

print("=" * 55)
print("EL RETO DEL ESPACIO QUÍMICO")
print("=" * 55)
print(f"Moléculas drug-like posibles:  10^60 (estimado)")
print(f"Moléculas en ChEMBL:           {moleculas_en_chembl:,}")
print(f"Fracción cubierta por ChEMBL:  {fraccion:.2e}")
print()
print("Es como buscar una aguja en un pajar del tamaño")
print("de 10^51 galaxias observables.")
print()
print("Por eso necesitamos inteligencia artificial:")
print("para navegar ese espacio de forma inteligente.")
````

## 2. El pipeline de descubrimiento de fármacos

El proceso de llevar un fármaco al mercado sigue etapas bien definidas.  
Este curso se enfoca principalmente en las **etapas tempranas** — donde los datos y la IA tienen más impacto.

```
┌─────────────────────────────────────────────────────────────────────────────────────┐
│                    PIPELINE DE DESCUBRIMIENTO DE FÁRMACOS                           │
├──────────────┬──────────────┬──────────────┬──────────────┬──────────────┬──────────┤
│  TARGET ID   │  HIT FINDING │  LEAD OPT.   │ PRECLÍNICO   │  CLÍNICO     │APROBACIÓN│
│              │              │              │              │  (Fases I-III│          │
│ ¿A qué       │ Encontrar    │ Mejorar la   │ Seguridad en │ Seguridad y  │ Agencias │
│ proteína     │ moléculas    │ molécula:    │ animales,    │ eficacia en  │ FDA,EMA, │
│ apuntamos?   │ activas      │ potencia,    │ ADME, tox.   │ humanos      │ INVIMA   │
│              │ (hits)       │ selectividad │              │              │          │
│   DATOS      │   DATOS      │   DATOS      │              │              │          │
│   IA         │   IA         │   IA         │              │              │          │
├──────────────┴──────────────┴──────────────┴──────────────┴──────────────┴──────────┤
│  ◄────────────── Este curso ──────────────►                                         │
│                                                           ◄── 10–15 años ──────────►│
└─────────────────────────────────────────────────────────────────────────────────────┘
```

**¿Qué hacemos en cada etapa temprana?**

| Etapa | Pregunta clave | ¿Cómo ayuda la IA? |
|-------|---------------|---------------------|
| **Target ID** | ¿Qué proteína causa la enfermedad? | Análisis de redes biológicas, genómica |
| **Hit Finding** | ¿Qué moléculas se unen al target? | Virtual screening, modelos QSAR |
| **Lead Optimization** | ¿Cómo mejoramos la molécula? | Química generativa, predicción de propiedades |

## 4. Caso real #2: Nirmatrelvir (Paxlovid) — IA en tiempo de pandemia

### El contexto
En diciembre de 2019, emergió el SARS-CoV-2. Para marzo de 2020 era pandemia global.  
El mundo necesitaba un antiviral oral efectivo — y lo necesitaba **rápido**.

### El target
La **proteasa principal del SARS-CoV-2** (Mpro, también llamada 3CL-proteasa) es esencial para la replicación viral. Sin ella, el virus no puede producir sus proteínas funcionales. Es un target atractivo porque:
- Es altamente conservada entre coronavirus (SARS-CoV-1 y SARS-CoV-2 comparten el 100% del sitio activo)
- No tiene homólogo humano cercano → menor toxicidad esperada
- Se conocía su estructura cristalográfica desde el brote de SARS-2002

### ¿Qué aprendemos de este caso?
1. **Los datos previos salvan vidas**: el conocimiento acumulado desde 2002 fue clave en 2020.

````python
# ── Comparación: pipeline clásico vs pipeline con IA ────────────────────────

timeline = {
    "Fase": [
        "Target ID",
        "Hit Finding",
        "Lead Optimization",
        "Preclínico",
        "Clínico (Fases I-III)",
        "Aprobación"
    ],
    "Pipeline clásico (años)": [2.0, 2.0, 3.0, 2.0, 8.0, 1.0],
    "Con IA/Computacional (años)": [0.5, 0.5, 1.0, 1.5, 6.0, 0.5]
}

print("IMPACTO DE LA IA EN EL PIPELINE")
print("=" * 62)
print(f"{'Fase':<25} {'Clásico':>10} {'Con IA':>10} {'Ahorro':>10}")
print("-" * 62)

total_clasico = 0
total_ia = 0

for i in range(len(timeline["Fase"])):
    fase = timeline["Fase"][i]
    clasico = timeline["Pipeline clásico (años)"][i]
    ia = timeline["Con IA/Computacional (años)"][i]
    ahorro = clasico - ia
    total_clasico += clasico
    total_ia += ia
    print(f"{fase:<25} {clasico:>8.1f}a {ia:>8.1f}a {ahorro:>+8.1f}a")

print("-" * 62)
print(f"{'TOTAL':<25} {total_clasico:>8.1f}a {total_ia:>8.1f}a {total_clasico-total_ia:>+8.1f}a")
print()
print(f"Reducción potencial: {((total_clasico-total_ia)/total_clasico)*100:.0f}% del tiempo total")
print()
print("Nota: Las estimaciones son aproximadas y varían según la")
print("indicación, la empresa y el estado del arte computacional.")
````

## 5. ChEMBL: la fuente de datos del curso

**ChEMBL** es una base de datos pública de bioactividad mantenida por el European Bioinformatics Institute (EMBL-EBI).  
Es la principal fuente de datos para construir modelos QSAR y de machine learning en química medicinal.

### ¿Qué contiene?

| Contenido | Cantidad aproximada (2024) |
|-----------|--------------------------|
| Moléculas (compuestos) | ~2.4 millones |
| Targets (proteínas, receptores) | ~15.000 |
| Actividades biológicas registradas | ~20 millones |
| Documentos científicos fuente | ~90.000 |

### ¿Qué es una "actividad biológica" en ChEMBL?
Cada registro conecta una **molécula** con un **target** a través de una **medida de actividad**:

```
Molécula (SMILES) ──→ Ensayo biológico ──→ Target (proteína) 
        ↓                                        ↓
   IC50 = 45 nM                           EGFR (Homo sapiens)
```

**Tipos de actividad más comunes:**
- **IC50**: Concentración que inhibe el 50% de la actividad del target
- **Ki**: Constante de inhibición (afinidad termodinámica)
- **EC50**: Concentración efectiva al 50% (para agonistas)
- **% inhibición**: Porcentaje de inhibición a una concentración fija

En la **semana 3** aprenderemos por qué estos tipos de datos **no son directamente comparables** y cómo curarlos correctamente.

### Paso 3: Buscar el target de imatinib — BCR-ABL

Vamos a replicar (de forma simplificada) el tipo de consulta que un científico computacional haría al inicio de un proyecto de drug discovery.


````python
# ── Buscar el target BCR-ABL en ChEMBL ─────────────────────────────────────
# new_client.target es el "buscador" de proteínas/targets en ChEMBL

target = new_client.target

# Buscamos por nombre — como si fuéramos al buscador de ChEMBL en la web
resultados_target = target.search('BCR-ABL')

# Convertimos a una tabla de pandas (DataFrame)
df_targets = pd.DataFrame.from_records(resultados_target)

print(f"Targets encontrados para 'BCR-ABL': {len(df_targets)}")
print()

# Mostramos las columnas más relevantes
columnas = ['target_chembl_id', 'pref_name', 'target_type', 'organism']
# Solo mostramos columnas que existen en el resultado
columnas_disponibles = [c for c in columnas if c in df_targets.columns]
print(df_targets[columnas_disponibles].head(10).to_string(index=False))
````

### Paso 5: Obtener las bioactividades registradas para BCR-ABL

Aquí es donde aparecen los datos de actividad de imatinib y otros inhibidores.


````python
# ── Consultar bioactividades de BCR-ABL ─────────────────────────────────────
activity = new_client.activity

# Filtramos por:
#   - target_chembl_id: nuestro target (BCR-ABL)
#   - standard_type: IC50 (el tipo de medida más común y comparable)
#   - assay_type: B = ensayo de unión (Binding assay)

bioactividades = activity.filter(
    target_chembl_id=TARGET_ID,
    standard_type='IC50',
    assay_type='B'
)

# Convertimos a DataFrame
df_activ = pd.DataFrame.from_records(bioactividades)

print(f"Registros de IC50 encontrados para BCR-ABL: {len(df_activ)}")
print()
print("Primeras 5 columnas disponibles:")
print([c for c in df_activ.columns[:8]])
````

````python
# ── Seleccionar y mostrar las columnas más informativas ─────────────────────
columnas_interes = [
    'molecule_chembl_id',    # ID de la molécula
    'canonical_smiles',      # Estructura en formato SMILES
    'standard_value',        # Valor numérico de IC50
    'standard_units',        # Unidades (nM, µM, etc.)
    'pchembl_value',         # pIC50 = -log10(IC50 en M) — más comparable
    'assay_chembl_id',       # ID del ensayo
    'document_year'          # Año de publicación
]

# Filtramos solo las columnas que existen en nuestro DataFrame
cols_disponibles = [c for c in columnas_interes if c in df_activ.columns]
df_vista = df_activ[cols_disponibles].copy()

# Convertimos el valor de IC50 a número (viene como texto desde la API)
if 'standard_value' in df_vista.columns:
    df_vista['standard_value'] = pd.to_numeric(df_vista['standard_value'], errors='coerce')

if 'pchembl_value' in df_vista.columns:
    df_vista['pchembl_value'] = pd.to_numeric(df_vista['pchembl_value'], errors='coerce')

print("MUESTRA DE DATOS DE IC50 PARA BCR-ABL")
print("=" * 70)
print(df_vista.head(8).to_string(index=False))
print()
print(f"Total de registros: {len(df_vista)}")
````

Busquemos específicamente los datos de actividad de imatinib contra BCR-ABL.


````python
# ── Buscar imatinib por nombre en ChEMBL ────────────────────────────────────
molecule = new_client.molecule

# Buscamos imatinib por su nombre preferido
resultado_imatinib = molecule.search('imatinib')
df_imatinib_mol = pd.DataFrame.from_records(resultado_imatinib)

# Encontrar el ChEMBL ID de imatinib
if len(df_imatinib_mol) > 0:
    # Buscamos la entrada con nombre exacto
    if 'pref_name' in df_imatinib_mol.columns:
        mask = df_imatinib_mol['pref_name'].str.upper() == 'IMATINIB'
        if mask.any():
            imatinib_id = df_imatinib_mol[mask]['molecule_chembl_id'].values[0]
            print(f"✅ Imatinib encontrado en ChEMBL")
            print(f"   ChEMBL ID: {imatinib_id}")
        else:
            # Tomar el primero
            print('Imatinib no encontrado en ChEMBL')
    
    # Ver sus propiedades si están disponibles
    if 'molecule_properties' in df_imatinib_mol.columns:
        props = df_imatinib_mol[mask]['molecule_properties'].values[0]
        if props:
            print()
            print("PROPIEDADES FISICOQUÍMICAS DE IMATINIB")
            print("-" * 40)
            props_interes = {
                'mw_freebase': 'Peso molecular (Da)',
                'alogp': 'LogP (lipofilia)',
                'hbd': 'Donadores de H-bond',
                'hba': 'Aceptores de H-bond',
                'psa': 'Área polar superficial (Å²)',
                'ro5_violations': 'Violaciones Regla de 5 (Lipinski)'
            }
            for key, label in props_interes.items():
                if key in props and props[key] is not None:
                    print(f"  {label:<35} {props[key]}")
````

````python
# ── Actividades de imatinib contra BCR-ABL ──────────────────────────────────
# Buscamos directamente por molécula y target

actividades_imatinib = activity.filter(
    target_chembl_id=TARGET_ID,
    molecule_chembl_id='CHEMBL941'    # ID de imatinib en ChEMBL
)

df_ima_activ = pd.DataFrame.from_records(actividades_imatinib)

if len(df_ima_activ) > 0:
    print("ACTIVIDADES DE IMATINIB CONTRA BCR-ABL EN ChEMBL")
    print("=" * 55)
    
    cols_show = ['standard_type', 'standard_value', 'standard_units', 
                 'pchembl_value', 'assay_description', 'document_year']
    cols_show = [c for c in cols_show if c in df_ima_activ.columns]
    
    for _, row in df_ima_activ[cols_show].head(10).iterrows():
        print(f"  Tipo:   {row.get('standard_type', 'N/A')}")
        print(f"  Valor:  {row.get('standard_value', 'N/A')} {row.get('standard_units', '')}")
        if 'pchembl_value' in row and pd.notna(row['pchembl_value']):
            print(f"  pIC50:  {row['pchembl_value']}")
        if 'document_year' in row:
            print(f"  Año:    {row.get('document_year', 'N/A')}")
        print()
else:
    print("No se encontraron actividades con ese filtro específico.")
    print("Prueba cambiando el ChEMBL ID de imatinib.")
````
